# House Prices — Step 2: Label Encoding

Label Encoding converts each category into an integer (0, 1, 2, ...). It's simple and memory-efficient, but it silently implies an **order** between categories (2 > 1 > 0).

That's exactly right for a genuinely **ordinal** column (e.g. quality ratings: Poor < Fair < Typical < Good < Excellent) *if* we map the integers in the correct order ourselves. It's **misleading** for a **nominal** column (e.g. `Neighborhood`) where categories have no real ranking — a generic encoder would just assign integers alphabetically, and a model could wrongly learn that `Neighborhood=24` is somehow "more" than `Neighborhood=1`.

This notebook demonstrates both cases, then applies Label Encoding across all remaining categorical columns to produce a fully numeric dataset.

## 1. Load the cleaned dataset

We continue from `train_cleaned.csv`, produced in Step 1 (Data Cleaning).

```python
import pandas as pd
pd.set_option('display.max_columns', 15)

df = pd.read_csv('train_cleaned.csv')
print("Shape:", df.shape)
df.head(3)
```

## 2. Ordinal example: `KitchenQual`

`KitchenQual` has a real, meaningful order: `Po` (Poor) < `Fa` (Fair) < `TA` (Typical/Average) < `Gd` (Good) < `Ex` (Excellent).

For ordinal columns like this, we define the mapping **manually** rather than letting a generic encoder guess — this way the integers actually reflect quality rank, which is useful information for a model.

```python
quality_map = {'Po': 0, 'Fa': 1, 'TA': 2, 'Gd': 3, 'Ex': 4}

df[['KitchenQual']].assign(KitchenQual_encoded=df['KitchenQual'].map(quality_map)).head(8)
```

## 3. Nominal example: `Neighborhood` — why generic Label Encoding is risky here

`Neighborhood` has 25 categories with **no natural order**. If we run sklearn's `LabelEncoder` on it, it just assigns integers alphabetically — `Blmngtn` might become `0` and `Veenker` might become `24`, with nothing meaningful behind that ranking.

We show it below **for illustration**, not because it's the right choice for this column (One-Hot or Target Encoding, covered in later notebooks, are better fits for high-cardinality nominal columns like this one).

```python
from sklearn.preprocessing import LabelEncoder

le_demo = LabelEncoder()
demo = df[['Neighborhood']].copy()
demo['Neighborhood_encoded'] = le_demo.fit_transform(demo['Neighborhood'])
demo.drop_duplicates().sort_values('Neighborhood_encoded').head(10)
```

## 4. Apply Label Encoding to all categorical columns

For this notebook's deliverable, we now encode **every** remaining categorical (text) column using `LabelEncoder`, so the output is a fully numeric dataframe. We keep a copy of each fitted encoder in a dictionary (`encoders`), in case we need to inverse-transform (recover the original category) later.

Note: as discussed above, this is a fast baseline approach (works well for tree-based models like Random Forest / XGBoost), but for nominal columns you may prefer One-Hot or Target Encoding, which we'll build in the next notebooks.

```python
categorical_cols = df.select_dtypes(include='object').columns.tolist()
print(f"Encoding {len(categorical_cols)} categorical columns")

df_label_encoded = df.copy()
encoders = {}

for col in categorical_cols:
    le = LabelEncoder()
    df_label_encoded[col] = le.fit_transform(df_label_encoded[col])
    encoders[col] = le

print("All columns now numeric:", (df_label_encoded.dtypes != 'object').all())
df_label_encoded.head(3)
```

## 5. Save the label-encoded dataset

We save this version separately from the cleaned dataset so we can compare it against the One-Hot and Target Encoding versions built in the next notebooks.

```python
df_label_encoded.to_csv('train_label_encoded.csv', index=False)
print("Saved train_label_encoded.csv with shape:", df_label_encoded.shape)
```
